## 3 — Feature Engineering

```python
train['year']       = train['Date'].dt.year
train['day_of_year'] = train['Date'].dt.dayofyear
train['month']      = train['Date'].dt.month
train['day']        = train['Date'].dt.day

# Annual totals — used to estimate YoY growth
annual = train.groupby('year')[['Revenue', 'COGS']].sum()
print('Annual totals (only complete years shown):')
print(annual)
```

```python
# 1. Additional Time Features
train['day_of_week']   = train['Date'].dt.dayofweek
train['quarter']       = train['Date'].dt.quarter
train['week_of_year'] = train['Date'].dt.isocalendar().week.astype(int)

# 2. Lag Features (Dịch chuyển dữ liệu)
for lag in [7, 14, 30, 365]:
    train[f'rev_lag_{lag}'] = train['Revenue'].shift(lag)
    train[f'cogs_lag_{lag}'] = train['COGS'].shift(lag)

# 3. Rolling Window Features (Trung bình trượt)
for window in [7, 30]:
    train[f'rev_roll_mean_{window}'] = train['Revenue'].transform(lambda x: x.rolling(window).mean())
    train[f'cogs_roll_mean_{window}'] = train['COGS'].transform(lambda x: x.rolling(window).mean())

# 4. Seasonal Flags
train['is_weekend']    = (train['day_of_week'] >= 5).astype(int)
train['is_month_end']  = train['Date'].dt.is_month_end.astype(int)
train['is_quarter_end'] = train['Date'].dt.is_quarter_end.astype(int)

# 5. Trend Feature (Số ngày kể từ điểm bắt đầu)
min_date = train['Date'].min()
train['trend_days'] = (train['Date'] - min_date).dt.days

print('Advanced features created. Shape:', train.shape)
display(train.tail())
```

```python
import os

TRAFFIC_FILE = DATA_DIR + 'web_traffic.csv'

if os.path.exists(TRAFFIC_FILE):
    # Đọc thử 1 vài dòng để kiểm tra tên cột
    df_check = pd.read_csv(TRAFFIC_FILE, nrows=0)
    cols = df_check.columns.tolist()
    print(f'Columns in web_traffic.csv: {cols}')

    # Tìm cột ngày tháng (không phân biệt hoa thường)
    date_col = next((c for c in cols if c.lower() == 'date'), None)

    if date_col:
        traffic = pd.read_csv(TRAFFIC_FILE, parse_dates=[date_col])
        # Chuẩn hóa tên cột về 'Date' để merge
        traffic = traffic.rename(columns={date_col: 'Date'})

        print('Traffic data shape:', traffic.shape)

        # Merge với train
        train = train.merge(traffic, on='Date', how='left')

        # Fill missing values với 0
        traffic_cols = [c for c in traffic.columns if c != 'Date']
        train[traffic_cols] = train[traffic_cols].fillna(0)

        print(f'Successfully merged {len(traffic_cols)} traffic columns.')
        display(train.tail())
    else:
        print(f'Error: Could not find a date column in {cols}')
else:
    print(f'Warning: {TRAFFIC_FILE} not found.')
```

```python
PROMO_FILE = DATA_DIR + 'promotions.csv'

if os.path.exists(PROMO_FILE):
    promo = pd.read_csv(PROMO_FILE, parse_dates=['start_date', 'end_date'])

    # Tạo cờ is_promo cho train
    train['is_promo'] = 0
    for idx, row in promo.iterrows():
        mask = (train['Date'] >= row['start_date']) & (train['Date'] <= row['end_date'])
        train.loc[mask, 'is_promo'] = 1

    print(f"Successfully added promotion flags. Days with promotions: {train['is_promo'].sum()}")
    display(train[['Date', 'is_promo']].tail())
else:
    print(f"Warning: {PROMO_FILE} not found.")
```

```python
REVIEWS_FILE = DATA_DIR + 'reviews.csv'

if os.path.exists(REVIEWS_FILE):
    reviews = pd.read_csv(REVIEWS_FILE, parse_dates=['review_date'])
    # Tính trung bình rating theo ngày
    daily_reviews = reviews.groupby('review_date')['rating'].agg(['mean', 'count']).reset_index()
    daily_reviews.columns = ['Date', 'avg_rating', 'review_count']

    # Merge vào train
    train = train.merge(daily_reviews, on='Date', how='left')
    train['avg_rating'] = train['avg_rating'].fillna(train['avg_rating'].mean())
    train['review_count'] = train['review_count'].fillna(0)

    print("Successfully added review features.")
    display(train[['Date', 'avg_rating', 'review_count']].tail())
else:
    print(f"Warning: {REVIEWS_FILE} not found.")
```

## 5 — Model Training (LightGBM)
Training two separate models for Revenue and COGS using the features engineered above.

```python
import lightgbm as lgb

# Loại bỏ reviews, chỉ giữ lại is_promo
drop_cols = ['Date', 'Revenue', 'COGS', 'rev_norm', 'cogs_norm', 'traffic_source', \
             'sessions', 'unique_visitors', 'page_views', 'bounce_rate', 'avg_session_duration_sec', \
             'avg_rating', 'review_count']
features = [c for c in train.columns if c not in drop_cols]

# Re-split data
train_set = train[train['year'] <= 2020].copy()
val_set   = train[train['year'].isin([2021, 2022])].copy()

train_clean = train_set.dropna()
val_clean   = val_set.dropna()

X_train, y_rev_train, y_cogs_train = train_clean[features], train_clean['Revenue'], train_clean['COGS']
X_val,   y_rev_val,   y_cogs_val   = val_clean[features],   val_clean['Revenue'],   val_clean['COGS']

best_params = {'learning_rate': 0.01, 'n_estimators': 1000, 'num_leaves': 20, 'random_state': 42}

print(f'Training with features: {features}')

# 1. Model for Revenue
model_rev = lgb.LGBMRegressor(**best_params)
model_rev.fit(
    X_train, y_rev_train,
    eval_set=[(X_val, y_rev_val)],
    callbacks=[lgb.early_stopping(100), lgb.log_evaluation(100)]
)

# 2. Model for COGS
model_cogs = lgb.LGBMRegressor(**best_params)
model_cogs.fit(
    X_train, y_cogs_train,
    eval_set=[(X_val, y_cogs_val)],
    callbacks=[lgb.early_stopping(100), lgb.log_evaluation(100)]
)
```

## 5.2 — Ensemble Modeling (LightGBM + XGBoost)
Kết hợp nhiều mô hình để cải thiện độ chính xác và giảm MAE.

```python
import xgboost as xgb
from sklearn.metrics import mean_absolute_error

# Khởi tạo tham số cho XGBoost
xgb_params = {
    'n_estimators': 1000,
    'learning_rate': 0.01,
    'max_depth': 5,
    'random_state': 42,
    'n_jobs': -1
}

print('Training XGBoost models with promotion feature...')
# 1. XGBoost cho Revenue
model_rev_xgb = xgb.XGBRegressor(**xgb_params)
model_rev_xgb.fit(X_train, y_rev_train, eval_set=[(X_val, y_rev_val)], verbose=False)

# 2. XGBoost cho COGS
model_cogs_xgb = xgb.XGBRegressor(**xgb_params)
model_cogs_xgb.fit(X_train, y_cogs_train, eval_set=[(X_val, y_cogs_val)], verbose=False)

# Dự báo Ensemble (Trung bình cộng 50-50)
rev_pred_lgb = model_rev.predict(X_val)
rev_pred_xgb = model_rev_xgb.predict(X_val)
rev_ensemble = (rev_pred_lgb + rev_pred_xgb) / 2

cogs_pred_lgb = model_cogs.predict(X_val)
cogs_pred_xgb = model_cogs_xgb.predict(X_val)
cogs_ensemble = (cogs_pred_lgb + cogs_pred_xgb) / 2

print('--- Ensemble Evaluation (Only Promo, No Reviews) ---')
print(f'New Revenue MAE: {mean_absolute_error(y_rev_val, rev_ensemble):,.2f}')
print(f'New COGS MAE:    {mean_absolute_error(y_cogs_val, cogs_ensemble):,.2f}')
```

## 6 — Model Evaluation (LightGBM)
Calculating MAE, RMSE, and R² scores on the validation set.

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Dự báo trên tập validation
rev_pred = model_rev.predict(X_val)
cogs_pred = model_cogs.predict(X_val)

def evaluate(y_true, y_pred, name):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    r2 = r2_score(y_true, y_pred)
    print(f'--- {name} Evaluation ---')
    print(f'MAE:  {mae:,.2f}')
    print(f'RMSE: {rmse:,.2f}')
    print(f'R2:   {r2:.4f}\n')

evaluate(y_rev_val, rev_pred, 'Revenue')
evaluate(y_cogs_val, cogs_pred, 'COGS')

# Plotting Actual vs Predicted
plt.figure(figsize=(15, 6))
plt.plot(val_clean['Date'], y_rev_val, label='Actual Revenue', alpha=0.7)
plt.plot(val_clean['Date'], rev_pred, label='Predicted Revenue', linestyle='--', color='red')
plt.title('LightGBM: Actual vs Predicted Revenue (Validation Set)')
plt.legend()
plt.show()
```

1. **R² (R-squared) ~ 0.80:** Con số này cho biết mô hình giải thích được **80% sự biến động** của dữ liệu. Trong dự báo chuỷi thờị gian thực tế, đây là mức độ chính xác rất tốt, cho thấy các đặc trưng như Lag và Rolling Mean đang hoạt động hiệu quả.

3. **RMSE (Root Mean Squared Error):** Tương tự như MAE nhưng phạt nặng hơn các sai số lồn. RMSE không quá chénh lệch vớị MAE chứng tỏ mô hình đồng nhất và ít bị ảnh hưởng bởị các giá trị ngoại lai (outliers).

### Dự báo cuốn chiếu (Recursive Forecasting) là gì?

**Khái niệm:** Dự báo cuốn chiếu là phương pháp dự báo cho tương lai mà trong đó, kết quả dự báo của bước trước (ví dụ: ngày hôm trước) được sử dụng làm đầu vào để tính toán các đặc trưng cho bước dự báo tiếp theo.

**Vì sao phải sử dụng trong bài toán này?**
1. **Đặc trưng phụ thuộc (Lag/Rolling Features):** Mô hình LightGBM của chúng ta được huấn luyện với các đặc trưng như `rev_lag_7` (doanh thu 7 ngày trước) hay `rev_roll_mean_7` (trung bình doanh thu 7 ngày gần nhất).
2. **Khoảng trống dữ liệu:** Khi bắt đầu dự báo cho ngày `2023-01-08`, chúng ta cần biết doanh thu của ngày `2023-01-01`. Tuy nhiên, ngày `2023-01-01` cũng là một ngày trong tương lai (không có trong tập Train).
3. **Giải pháp:** Chúng ta phải dự báo ngày 01 trước, lấy kết quả đó để tính toán đặc trưng cho ngày 08. Quá trình này lặp lại liên tục cho đến hết chu kỳ dự báo. Nếu không làm vậy, các đặc trưng Lag sẽ bị trống (NaN) và mô hình sẽ không thể đưa ra dự báo chính xác.

```python
import pandas as pd
import numpy as np

# 1. Chuẩn bị khung dữ liệu Test
test_df = test[['Date']].copy()
test_df['Revenue'] = np.nan
test_df['COGS'] = np.nan

# Kết hợp train và test
# Lấy tất cả các cột hiện có của train để tránh mất features (is_promo, avg_rating, v.v.)
full_df = pd.concat([train, test_df], axis=0).reset_index(drop=True)

# Gộp dữ liệu traffic vào full_df nếu cần (đã xử lý ở bước trước nhưng đảm bảo an toàn)
if 'traffic' in globals():
    # Chỉ merge nếu các cột traffic chưa tồn tại
    if 'sessions' not in full_df.columns:
        full_df = full_df.merge(traffic, on='Date', how='left')
    traffic_cols = [c for c in traffic.columns if c != 'Date']
    full_df[traffic_cols] = full_df[traffic_cols].fillna(0)

# Điền giá trị mặc định cho các feature mới ở giai đoạn test (2023-2024)
full_df['is_promo'] = full_df['is_promo'].fillna(0)
full_df['avg_rating'] = full_df['avg_rating'].fillna(train['avg_rating'].mean())
full_df['review_count'] = full_df['review_count'].fillna(0)

# 2. Dự báo cuốn chiếu từng ngày bằng LightGBM
for i in range(len(train), len(full_df)):
    current_date = full_df.loc[i, 'Date']

    # Tính toán features thời gian
    full_df.loc[i, 'year'] = current_date.year
    full_df.loc[i, 'day_of_year'] = current_date.dayofyear
    full_df.loc[i, 'month'] = current_date.month
    full_df.loc[i, 'day'] = current_date.day
    full_df.loc[i, 'day_of_week'] = current_date.dayofweek
    full_df.loc[i, 'quarter'] = current_date.quarter
    full_df.loc[i, 'week_of_year'] = int(current_date.isocalendar()[1])
    full_df.loc[i, 'is_weekend'] = 1 if current_date.dayofweek >= 5 else 0
    full_df.loc[i, 'is_month_end'] = 1 if current_date.is_month_end else 0
    full_df.loc[i, 'is_quarter_end'] = 1 if current_date.is_quarter_end else 0
    full_df.loc[i, 'trend_days'] = (current_date - train['Date'].min()).days

    # Cập nhật Lag features từ kết quả đã dự báo
    for lag in [7, 14, 30, 365]:
        full_df.loc[i, f'rev_lag_{lag}'] = full_df.loc[i-lag, 'Revenue'] if i >= lag else 0
        full_df.loc[i, f'cogs_lag_{lag}'] = full_df.loc[i-lag, 'COGS'] if i >= lag else 0

    # Cập nhật Rolling features
    for window in [7, 30]:
        full_df.loc[i, f'rev_roll_mean_{window}'] = full_df.loc[i-window:i-1, 'Revenue'].mean()
        full_df.loc[i, f'cogs_roll_mean_{window}'] = full_df.loc[i-window:i-1, 'COGS'].mean()

    # Lấy các features theo đúng danh sách đã huấn luyện
    X_curr = full_df.loc[[i], features].fillna(0)

    # Dự báo bằng model đã train
    full_df.loc[i, 'Revenue'] = model_rev.predict(X_curr)[0]
    full_df.loc[i, 'COGS'] = model_cogs.predict(X_curr)[0]

# Trích xuất kết quả dự báo ra biến final_submission
final_submission = full_df.iloc[len(train):][['Date', 'Revenue', 'COGS']].copy()
display(final_submission.head())
```

2. **Xác nhận xu hướng (Trend):** Kiểm tra xem mô hình có bắt được đúng bối cảnh hiện tại không (ví dụ: nếu doanh thu đang giảm ở cuối năm 2022 thì dự báo 2023 nén nối tiếp đà giảm đó).
3. **Độ tin cậy:** Giúp ngườ dùng tự tin rằng kết quả này có thể sử dụng để lập kế hoạch kinh doanh thực tế.
