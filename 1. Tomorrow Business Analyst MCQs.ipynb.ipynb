```python
import pandas as pd

# 1. Đọc dữ liệu từ file
df = pd.read_csv('orders.csv')

# 2. Chuyển cột ngày tháng sang định dạng chuẩn (datetime) để có thể cộng trừ
df['order_date'] = pd.to_datetime(df['order_date'])

# 3. Sắp xếp lại toàn bộ bảng: Nhóm các đơn của cùng 1 khách hàng đứng cạnh nhau,
# và xếp theo thứ tự thời gian từ cũ đến mới.
df = df.sort_values(by=['customer_id', 'order_date'])

# 4. Tính số ngày khoảng cách (inter-order gap)
# Hàm diff() sẽ lấy ngày của dòng dưới trừ đi ngày của dòng trên.
# Vì đã nhóm theo customer_id, nó sẽ tính đúng khoảng cách giữa 2 lần mua của cùng 1 người.
# Với đơn hàng đầu tiên của khách, phép tính sẽ ra NaN (không có đơn trước đó để trừ).
df['inter_order_gap'] = df.groupby('customer_id')['order_date'].diff().dt.days

# 5. Tính trung vị (median)
# Hàm median() sẽ tự động bỏ qua các giá trị NaN và tính trung vị của các khoảng cách ngày.
median_gap = df['inter_order_gap'].median()

print(f"Đáp án Câu 1 - Trung vị số ngày giữa hai lần mua liên tiếp là: {median_gap}")
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ file products.csv
df_products = pd.read_csv('products.csv')

# 2. Tạo một cột mới tên là 'gross_margin' để tính tỷ suất lợi nhuận gộp
# Áp dụng đúng công thức đề cho: (price - cogs) / price
df_products['gross_margin'] = (df_products['price'] - df_products['cogs']) / df_products['price']

# 3. Nhóm dữ liệu theo phân khúc (segment) và tính giá trị trung bình (mean) của cột gross_margin
avg_margin_by_segment = df_products.groupby('segment')['gross_margin'].mean()

# 4. Tìm phân khúc có giá trị trung bình cao nhất
best_segment = avg_margin_by_segment.idxmax()
best_margin_value = avg_margin_by_segment.max()

print("--- KẾT QUẢ CÂU 2 ---")
print(f"Phân khúc có tỷ suất LN trung bình cao nhất: {best_segment}")
print(f"Giá trị trung bình cụ thể là: {best_margin_value:.2%}")

# Dòng này in thêm danh sách tất cả các segment để bạn dễ đối chiếu với 4 đáp án A, B, C, D
print("\nChi tiết các phân khúc:")
print(avg_margin_by_segment)
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ 2 file
df_products = pd.read_csv('products.csv')
df_returns = pd.read_csv('returns.csv')

# 2. Join (kết nối) 2 bảng lại với nhau dựa trên cột chung là 'product_id'
# Hàm merge sẽ tự động ghép các dòng có cùng mã sản phẩm lại với nhau
df_merged = pd.merge(df_returns, df_products, on='product_id')

# 3. Lọc ra các sản phẩm thuộc danh mục 'Streetwear'
streetwear_returns = df_merged[df_merged['category'] == 'Streetwear']

# 4. Đếm số lần xuất hiện của từng lý do trả hàng (return_reason)
reason_counts = streetwear_returns['return_reason'].value_counts()

# Lấy lý do đứng top 1
top_reason = reason_counts.idxmax()

print("--- KẾT QUẢ CÂU 3 ---")
print(f"Lý do trả hàng xuất hiện nhiều nhất đối với Streetwear là: {top_reason}")
print("\nChi tiết số lượng các lý do:")
print(reason_counts)
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ file web_traffic.csv
df_web = pd.read_csv('web_traffic.csv')

# 2. Nhóm dữ liệu theo nguồn truy cập (traffic_source)
# và tính giá trị trung bình (mean) của cột tỷ lệ thoát (bounce_rate)
avg_bounce_by_source = df_web.groupby('traffic_source')['bounce_rate'].mean()

# 3. Tìm nguồn có tỷ lệ thoát trung bình thấp nhất
lowest_bounce_source = avg_bounce_by_source.idxmin()
lowest_bounce_value = avg_bounce_by_source.min()

print("--- KẾT QUẢ CÂU 4 ---")
print(f"Nguồn truy cập có tỷ lệ thoát trung bình thấp nhất: {lowest_bounce_source}")
print(f"Giá trị cụ thể là: {lowest_bounce_value:.5f}")

# In ra toàn bộ danh sách đã được sắp xếp từ thấp đến cao để dễ đối chiếu với 4 đáp án A, B, C, D
print("\nBảng xếp hạng tỷ lệ thoát trung bình của tất cả các nguồn:")
print(avg_bounce_by_source.sort_values())
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ file order_items.csv
df_items = pd.read_csv('order_items.csv')

# 2. Tính tỷ lệ phần trăm
# Hàm notnull() sẽ biến cột promo_id thành một cột mới gồm True (nếu có mã) và False (nếu null).
# Trong Python, True = 1 và False = 0.
# Do đó, dùng hàm mean() tính trung bình cộng của toàn bộ cột này sẽ ra chính xác tỷ lệ phần trăm (dưới dạng số thập phân).
promo_percentage = df_items['promo_id'].notnull().mean() * 100

print("--- KẾT QUẢ CÂU 5 ---")
print(f"Tỷ lệ các dòng có áp dụng khuyến mãi xấp xỉ là: {promo_percentage:.0f}%")
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ 2 file
df_customers = pd.read_csv('customers.csv')
df_orders = pd.read_csv('orders.csv')

# 2. Lọc bỏ các khách hàng có age_group là null
valid_customers = df_customers[df_customers['age_group'].notnull()]

# 3. Tính MẪU SỐ: Tổng số lượng khách hàng trong mỗi nhóm tuổi
customers_per_group = valid_customers.groupby('age_group')['customer_id'].nunique()

# 4. Join bảng khách hàng (đã lọc) với bảng đơn hàng để tính TỬ SỐ
merged_df = pd.merge(valid_customers[['customer_id', 'age_group']], df_orders[['order_id', 'customer_id']], on='customer_id')

# Tính TỬ SỐ: Tổng số đơn hàng tạo bởi mỗi nhóm tuổi
orders_per_group = merged_df.groupby('age_group')['order_id'].nunique()

# 5. Tính trung bình số đơn trên mỗi khách hàng (Tử số / Mẫu số)
avg_orders_per_customer = orders_per_group / customers_per_group

# 6. Tìm nhóm tuổi có con số trung bình cao nhất
top_age_group = avg_orders_per_customer.idxmax()
top_value = avg_orders_per_customer.max()

print("--- KẾT QUẢ CÂU 6 ---")
print(f"Nhóm tuổi có số đơn hàng trung bình cao nhất là: {top_age_group}")
print(f"Số đơn trung bình của nhóm này là: {top_value:.2f} đơn/khách")

print("\nChi tiết số đơn trung bình của tất cả các nhóm tuổi:")
print(avg_orders_per_customer.sort_values(ascending=False))
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ 3 file chi tiết
df_items = pd.read_csv('order_items.csv')
df_orders = pd.read_csv('orders.csv')
df_geo = pd.read_csv('geography.csv')

# 2. Tính doanh thu thực tế cho từng dòng sản phẩm
# Doanh thu = (Số lượng * Đơn giá) - Tiền giảm giá. Điền số 0 nếu discount_amount bị null.
df_items['discount_amount'] = df_items['discount_amount'].fillna(0)
df_items['item_revenue'] = (df_items['quantity'] * df_items['unit_price']) - df_items['discount_amount']

# 3. Gộp tổng doanh thu theo từng mã đơn hàng (order_id)
revenue_per_order = df_items.groupby('order_id')['item_revenue'].sum().reset_index()

# 4. Nối (Join) với bảng orders để lấy mã bưu chính (zip) của đơn hàng đó
merged_orders = pd.merge(revenue_per_order, df_orders[['order_id', 'zip']], on='order_id')

# 5. Nối tiếp với bảng địa lý (geography) để lấy tên vùng (region)
final_df = pd.merge(merged_orders, df_geo[['zip', 'region']], on='zip')

# 6. Tính tổng toàn bộ doanh thu theo từng vùng
revenue_by_region = final_df.groupby('region')['item_revenue'].sum()

# 7. Sắp xếp từ cao xuống thấp và in kết quả
print("--- KẾT QUẢ CÂU 7 ---")
print("Bảng xếp hạng doanh thu theo vùng:")
# In ra với định dạng có dấu phẩy ngăn cách hàng nghìn cho dễ nhìn
print(revenue_by_region.sort_values(ascending=False).apply(lambda x: f"{x:,.2f} VNĐ"))
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ file orders.csv
df_orders = pd.read_csv('orders.csv')

# 2. Lọc ra các đơn hàng có trạng thái bị huỷ
cancelled_orders = df_orders[df_orders['order_status'] == 'cancelled']

# 3. Đếm số lần xuất hiện của từng phương thức thanh toán trong tập dữ liệu vừa lọc
payment_counts = cancelled_orders['payment_method'].value_counts()

# 4. Tìm phương thức thanh toán đứng đầu (xuất hiện nhiều nhất)
top_payment = payment_counts.idxmax()
top_count = payment_counts.max()

print("--- KẾT QUẢ CÂU 8 ---")
print(f"Phương thức thanh toán được dùng nhiều nhất cho đơn huỷ là: {top_payment}")
print(f"Số lượng đơn huỷ của phương thức này: {top_count} đơn")

print("\nChi tiết bảng xếp hạng phương thức thanh toán của đơn huỷ:")
print(payment_counts)
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ 3 file
df_items = pd.read_csv('order_items.csv')
df_returns = pd.read_csv('returns.csv')
df_products = pd.read_csv('products.csv')

# 2. Đếm số dòng đã bán (trong order_items) theo từng size
# Ghép order_items với products (chỉ lấy cột product_id và size cho nhẹ)
items_with_size = pd.merge(df_items, df_products[['product_id', 'size']], on='product_id')
sales_count = items_with_size['size'].value_counts()

# 3. Đếm số bản ghi trả lại (trong returns) theo từng size
returns_with_size = pd.merge(df_returns, df_products[['product_id', 'size']], on='product_id')
returns_count = returns_with_size['size'].value_counts()

# 4. Tính tỷ lệ trả hàng
# Phép chia này của Pandas sẽ tự động gióng đúng dòng của size S chia cho size S, M chia cho M...
return_rate = returns_count / sales_count

# 5. Chỉ lấy 4 kích thước đề bài hỏi (S, M, L, XL) và tìm size cao nhất
target_sizes = ['S', 'M', 'L', 'XL']
# Lọc ra tập kết quả chỉ chứa 4 size này
final_rates = return_rate.loc[target_sizes]

top_size = final_rates.idxmax()
top_rate = final_rates.max()

print("--- KẾT QUẢ CÂU 9 ---")
print(f"Kích thước có tỷ lệ trả hàng cao nhất là: {top_size}")
print(f"Tỷ lệ cụ thể là: {top_rate:.2%}")

print("\nChi tiết tỷ lệ trả hàng của các kích thước:")
print(final_rates.apply(lambda x: f"{x:.2%}").sort_values(ascending=False))
```

```python
import pandas as pd

# 1. Đọc dữ liệu từ file payments.csv
df_payments = pd.read_csv('payments.csv')

# 2. Nhóm theo số kỳ trả góp (installments) và tính giá trị trung bình (mean)
avg_payment_by_installment = df_payments.groupby('installments')['payment_value'].mean()

# 3. Lọc ra 4 kỳ hạn cần xét để khớp với đáp án A, B, C, D
target_installments = [1, 3, 6, 12]

# Dùng .loc để chỉ lấy dữ liệu của 4 kỳ hạn này (phòng trường hợp data thực có thêm kỳ 2, 9, 24...)
# Thêm kiểm tra intersection để tránh lỗi nếu có kỳ hạn không tồn tại trong data
available_targets = avg_payment_by_installment.index.intersection(target_installments)
final_avg = avg_payment_by_installment.loc[available_targets]

# 4. Tìm kỳ hạn có giá trị thanh toán trung bình cao nhất
top_installment = final_avg.idxmax()
top_value = final_avg.max()

print("--- KẾT QUẢ CÂU 10 ---")
print(f"Kế hoạch trả góp có giá trị thanh toán trung bình cao nhất là: {top_installment} kỳ")
print(f"Giá trị trung bình cụ thể là: {top_value:,.2f}")

print("\nChi tiết giá trị trung bình của các kế hoạch trả góp:")
print(final_avg.apply(lambda x: f"{x:,.2f}").sort_values(ascending=False))
```
